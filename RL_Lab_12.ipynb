```python
# Value Iteration in a Deterministic Grid World

# Define states and actions
states = ["S0", "S1", "G"]
actions = ["Up", "Down", "Left", "Right"]

gamma = 0.9
V = {"S0": 0, "S1": 0, "G": 0}

# Define transitions and rewards
transitions = {
    "S0": {
        "Up": ("S0", -1),
        "Down": ("S1", -1),
        "Left": ("S0", -1),
        "Right": ("S0", -1)
    },
    "S1": {
        "Up": ("S0", -1),
        "Down": ("S1", -1),
        "Left": ("S1", -1),
        "Right": ("G", 100)
    }
}

# Display transition information
print("Transition Information:")
for state in ["S0", "S1"]:
    for action in actions:
        next_state, reward = transitions[state][action]
        print(state, action, next_state, reward)

# Value Iteration
iteration = 0
theta = 0.000001

print("\nIteration-wise State Values")

while True:
    iteration += 1
    new_V = V.copy()

    for state in ["S0", "S1"]:
        q_values = []

        for action in actions:
            next_state, reward = transitions[state][action]

            q = reward + gamma * V[next_state]
            q_values.append(q)

        new_V[state] = max(q_values)

    delta = max(abs(new_V[s] - V[s]) for s in ["S0", "S1"])

    V = new_V

    print("Iteration", iteration, ":", V)

    if delta < theta:
        break

# Calculate final Q values
print("\nFinal State Values:")
for state in states:
    print(state, "=", V[state])

print("\nFinal Q Values:")

Q = {}

for state in ["S0", "S1"]:
    Q[state] = {}

    for action in actions:
        next_state, reward = transitions[state][action]
        Q[state][action] = reward + gamma * V[next_state]

        print(state, action, "=", Q[state][action])

# Find optimal policy
print("\nOptimal Policy:")

for state in ["S0", "S1"]:
    best_action = max(Q[state], key=Q[state].get)
    print(state, ":", best_action)

print("\nOptimal Path: S0 -> S1 -> G")
```
